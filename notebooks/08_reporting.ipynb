# 8. Automated Reporting & Summary
Questo notebook illustra come estrarre e visualizzare il report testuale automatizzato generato dalla pipeline, fornendo una sintesi finale dello stato della caldera.

```python
import os
import pandas as pd
from IPython.display import Markdown, display

# Percorsi ai file di output elaborati dalla pipeline
report_path = os.path.abspath('../reports/weekly_report.md')
ews_path = os.path.abspath('../data/processed/early_warning_system.csv')
forecast_path = os.path.abspath('../data/processed/dl_forecast.csv')
```

## Visualizzazione del Report Settimanale
Se la pipeline principale (`run_pipeline.py`) o lo script `generate_report.py` sono stati eseguiti con successo, possiamo renderizzare il contenuto del report Markdown direttamente qui.

```python
if os.path.exists(report_path):
    with open(report_path, 'r', encoding='utf-8') as f:
        report_content = f.read()
    display(Markdown(report_content))
else:
    print("Nessun report trovato. Esegui 'python scripts/generate_report.py' per generarlo.")
```

## Analisi degli Ultimi Dati Operativi
Estrazione rapida degli ultimi indicatori registrati dall'Early Warning System e dal Modello di Forecasting AI per confermare visivamente le metriche discusse nel report testuale.

```python
if os.path.exists(ews_path):
    df_ews = pd.read_csv(ews_path)
    print("Ultimi 5 giorni registrati nel sistema di allerta (EWS):\n")
    display(df_ews[['time', 'seismic_rate', 'b_value', 'unrest_index', 'state', 'alert_flag']].tail(5))
else:
    print("Dati EWS non trovati. Eseguire l'analisi dell'Early Warning.")

if os.path.exists(forecast_path):
    df_fc = pd.read_csv(forecast_path)
    print("\nPrevisione AI per i prossimi 7 giorni (Modello LSTM):\n")
    display(df_fc)
```
