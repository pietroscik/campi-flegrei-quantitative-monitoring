# 5. Seismic Energy & B-Value
Analisi dello stress crostale tramite la legge di Gutenberg-Richter e calcolo dell'energia sismica cumulata (Proxy di Benioff).

```python
import sys, os
import pandas as pd
import matplotlib.pyplot as plt
sys.path.append(os.path.abspath('..'))
from src.analysis.b_value import rolling_b_value_time_based

cat_path = os.path.abspath('../data/processed/catalog_clean.csv')
if os.path.exists(cat_path):
    df = pd.read_csv(cat_path, parse_dates=['time'])
    
    b_df = rolling_b_value_time_based(df, window_days=180, min_events=50, dynamic_m0=True)
    
    # Energy Proxy: E ~ 10^(1.5 * M)
    df['energy'] = 10 ** (1.5 * df['magnitude'])
    df['cum_energy'] = df['energy'].cumsum()
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=False)
    ax1.plot(b_df['time'], b_df['b_value'], color='darkgreen', marker='o')
    ax1.set_title('Evoluzione temporale del b-value')
    ax2.plot(df['time'], df['cum_energy'], color='darkred')
    ax2.set_title('Energia Sismica Cumulata (Benioff Proxy)')
    plt.tight_layout()
    plt.show()
```
