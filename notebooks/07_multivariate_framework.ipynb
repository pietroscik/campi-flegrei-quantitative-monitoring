# 7. Multivariate Framework & Unrest Index
Esplorazione finale del sistema multi-segnale (Sismicità + B-Value) e visualizzazione dell'Early Warning System combinato generato dalla Pipeline.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

unrest_path = os.path.abspath('../data/processed/early_warning_system.csv')
if os.path.exists(unrest_path):
    df = pd.read_csv(unrest_path, parse_dates=['time'])
    
    fig, ax = plt.subplots(figsize=(14, 6))
    ax.plot(df['time'], df['unrest_index'], color='navy', label='Unrest Index', linewidth=2)
    
    if 'threshold_alert' in df.columns:
        ax.plot(df['time'], df['threshold_baseline'], 'g--', label='Baseline', alpha=0.6)
        ax.plot(df['time'], df['threshold_alert'], 'r-', label='Allarme Critico', alpha=0.6)
    
    alerts = df[df['alert_flag'] == 1]
    if not alerts.empty:
        ax.scatter(alerts['time'], alerts['unrest_index'], color='red', marker='^', s=100, label='Alert Attivato', zorder=5)
    
    plt.title('Campi Flegrei - Sistema Multi-Variato (Early Warning System)')
    plt.ylabel('Unrest Index (Z-Score)')
    plt.legend()
    plt.tight_layout()
    plt.show()
else:
    print('Esegui prima l\'intera pipeline per generare l\'Unrest Index.')
```
