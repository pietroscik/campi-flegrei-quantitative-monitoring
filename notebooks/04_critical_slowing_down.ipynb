# 4. Critical Slowing Down (CSD) Indicators
Valutazione della varianza e dell'autocorrelazione (Lag-1) sul rateo sismico. Aumenti prolungati di questi due valori preannunciano la perdita di resilienza del sistema vulcanico (approccio ai "tipping point").

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

cat_path = os.path.abspath('../data/processed/catalog_clean.csv')
if os.path.exists(cat_path):
    df = pd.read_csv(cat_path, parse_dates=['time'])
    daily_rate = df.set_index('time').resample('D').size()
    
    window = 90
    rolling_var = daily_rate.rolling(window).var()
    rolling_ar1 = daily_rate.rolling(window).apply(lambda x: pd.Series(x).autocorr(lag=1))
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
    ax1.plot(rolling_var, color='teal')
    ax1.set_title(f'Varianza Mobile del Rateo Sismico ({window} giorni)')
    ax2.plot(rolling_ar1, color='magenta')
    ax2.set_title(f'Autocorrelazione Lag-1 ({window} giorni) - Segnale CSD')
    plt.tight_layout()
    plt.show()
```
