# 6. SARIMA Modeling & Anomaly Detection
Applicazione del modello autoregressivo statistico SARIMA per modellare il background rate sismico ed estrarre le anomalie basate sui residui (i picchi non previsti dall'andamento base).

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from statsmodels.tsa.statespace.sarimax import SARIMAX

cat_path = os.path.abspath('../data/processed/catalog_clean.csv')
if os.path.exists(cat_path):
    df = pd.read_csv(cat_path, parse_dates=['time'])
    weekly_rate = df.set_index('time').resample('W').size()
    
    # Train SARIMA (1,1,1)
    model = SARIMAX(weekly_rate, order=(1, 1, 1), seasonal_order=(0, 0, 0, 0))
    results = model.fit(disp=False)
    
    weekly_rate.plot(label='Tasso Sismico Osservato', figsize=(12, 5), color='blue')
    results.fittedvalues.plot(label='Modello SARIMA', color='red', alpha=0.7)
    plt.title('Modellazione del Tasso Sismico')
    plt.legend()
    plt.show()
    
    residuals = results.resid
    anomalies = residuals[residuals > residuals.std() * 2]
    print(f"Anomalie strutturali rilevate (> 2 std): {len(anomalies)}")
```
