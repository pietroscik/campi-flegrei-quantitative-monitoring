# 2. Data Cleaning & Preprocessing
Filtriamo i duplicati, eliminiamo le magnitudo non valide (Negative o NaN) e standardizziamo il catalogo per renderlo utilizzabile dai modelli di Deep Learning e Statistica.

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
sys.path.append(os.path.abspath('..'))

from src.preprocessing.clean_catalog import build_catalog

raw_path = os.path.abspath('../data/raw/ingv_events.csv')
if os.path.exists(raw_path):
    df_clean = build_catalog(raw_path)
    df_clean['time'] = pd.to_datetime(df_clean['time'])
    
    plt.figure(figsize=(12, 5))
    df_clean.set_index('time').resample('M').size().plot(title='Eventi Sismici al mese (Catalogo Pulito)', color='navy')
    plt.ylabel('Eventi')
    plt.show()
else:
    print('Dati raw non trovati. Eseguire prima il notebook 01.')
```
