# 3. GNSS & Break Detection (Deformazione del Suolo)
Analisi del sollevamento bradisismico (Uplift) e rilevamento di rotture di trend strutturali. Poiché i dati GNSS dell'INGV non sono distribuiti via API pubbliche continue, simuliamo o estraiamo il trend per testare l'algoritmo strutturale.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Simulazione dati storici GNSS
dates = pd.date_range('2014-01-01', periods=120, freq='ME')
uplift = np.linspace(0, 100, 120) + np.random.normal(0, 1, 120)
df_gnss = pd.DataFrame({'time': dates, 'uplift_cm': uplift}).set_index('time')

# Rilevamento accelerazione (Break detection tramite derivata)
df_gnss['velocity'] = df_gnss['uplift_cm'].diff()

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
ax1.plot(df_gnss.index, df_gnss['uplift_cm'], color='purple', marker='.')
ax1.set_title('Sollevamento Bradisismico Cumulato (cm)')
ax2.bar(df_gnss.index, df_gnss['velocity'], color='orange', width=20)
ax2.axhline(df_gnss['velocity'].mean() + 2*df_gnss['velocity'].std(), color='red', linestyle='--', label='Anomalia di Velocità')
ax2.set_title('Velocità di Sollevamento (cm/mese)')
ax2.legend()
plt.tight_layout()
plt.show()
```
